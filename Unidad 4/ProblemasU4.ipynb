```python
from scipy.stats import norm

# Parámetros de la distribución normal
media = 70
desviacion_estandar = 4
num_estudiantes = 500

# Probabilidad de que un estudiante pese más de 68 y menos de 72 kg
probabilidad_entre_68_y_72 = norm.cdf(72, media, desviacion_estandar) - norm.cdf(68, media, desviacion_estandar)

print("1. Probabilidad de que un estudiante pese más de 68 y menos de 72 kg:", probabilidad_entre_68_y_72)

# Cantidad de estudiantes que pesan más de 68 y menos de 72 kg
cantidad_estudiantes = probabilidad_entre_68_y_72 * num_estudiantes

print("2. Cantidad de estudiantes que pesan más de 68 y menos de 72 kg:", cantidad_estudiantes)
```
