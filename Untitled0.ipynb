```python
from statistics import mean

grupoa = [70, 70, 100,95, 45, 80, 73, 20, 100, 96 ]   # ejemplo de lista de calificaciones
Promedio2 = mean(grupoa)
print(Promedio2)
```

```python
from statistics import median

# Determinar el valor de la media con la funcion median

mediana = median(grupoa)
print(mediana)
```

```python
from statistics import mode

# Determinar el valor de la moda con la funcion mode
moda = mode(grupoa)
print(moda)
```

```python
from statistics import multimode

# En caso de que exista mas de una moda, se puede emplear la funcion multimode para sacar
#Multiples modas que pueden existir

moda2 = multimode(grupoa)
print(moda2)
```

```python
tiempos100mts = [15,14,11,20,21,23,25,20,19,18]
print(tiempos100mts)
```
