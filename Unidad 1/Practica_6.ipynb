##### **Practica 6: Estadística descriptiva Dataset Autos**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Cargar el archivo CSV
df = pd.read_csv("autos2.csv")
```

```python
# Estadísticas descriptivas básicas y medidas de tendencia central de autos2
print("Mínimo:            ", df["price"].min())
print("Máximo:            ", df["price"].max())
print("Moda:              ", df["price"].mode().tolist())
print("Media:             ", df["price"].mean())
print("Mediana:           ", df["price"].median())
print("Desviación estándar:", df["price"].std())

df["price"].describe()
```

## 2. Historiagrama de los precios de los automoviles

```python
n = df["price"].count()                    # 201
k = int(np.ceil(1 + 3.322 * np.log10(n)))  # 9

plt.figure(figsize=(9, 5))
plt.hist(df["price"].dropna(), bins=k, color="#4C72B0", edgecolor="white")
plt.axvline(df["price"].mean(),   color="red",   linestyle="--",
            label=f"Media = ${df['price'].mean():,.0f}")
plt.axvline(df["price"].median(), color="green", linestyle="-.",
            label=f"Mediana = ${df['price'].median():,.0f}")
plt.title(f"Distribución de precios de automóviles (k = {k} bins, regla de Sturges)")
plt.xlabel("Precio (USD)")
plt.ylabel("Frecuencia (número de autos)")
plt.legend()
plt.show()
```

## 3.Realizar un diagrama de caja y bigotes que muestre la relación estadística entre la columna del precio del automóvil (price) y tipo de vehículo (body style)

```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x="body-style", y="price",
            order=["hatchback", "wagon", "sedan", "convertible", "hardtop"])
plt.title("Precio del automóvil según el tipo de carrocería (body-style)")
plt.xlabel("Tipo de carrocería")
plt.ylabel("Precio (USD)")
plt.show()
```

## 4. Realizar un diagrama de caja y bigotes que muestre la relación estadística entre la columna del precio del automóvil (price) y tipo tracción ruedas (drive wheels)

```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x="drive-wheels", y="price", order=["fwd", "4wd", "rwd"])
plt.title("Precio del automóvil según el tipo de tracción (drive-wheels)")
plt.xlabel("Tracción de ruedas")
plt.ylabel("Precio (USD)")
plt.show()
```

## 5. Realizar un diagrama de caja y bigotes que muestre la relación estadística entre la columna del precio del automóvil (price) y la ubicación del motor (engine-location)

#

```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x="engine-location", y="price", order=["front", "rear"])
plt.title("Precio del automóvil según la ubicación del motor (engine-location)")
plt.xlabel("Ubicación del motor")
plt.ylabel("Precio (USD)")
plt.show()

df["engine-location"].value_counts()
```
