```python
from sklearn.linear_model import LinearRegression
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

##Grupo 1

> Añadir blockquote



```python
#Cargar el grupo 1
grupo1 = pd.read_csv("grupo1.csv")
```

```python
#Mostrar las 2 columnas del archivo
grupo1.head()
```

```python
#Descripcion estadistica
print(grupo1.describe())

#Histograma de calificaciones
plt.figure(figsize=(6,4))
plt.hist(grupo1['Calificacion'], bins=10, edgecolor='black')
plt.title('Histograma de Calificaciones')
plt.xlabel('Calificacion')
plt.ylabel('Frecuencia')
plt.grid(True)
plt.show()

#Datos de entrada
horas = grupo1[['HorasEstudio']]
calificacion = grupo1['Calificacion']

modelo = LinearRegression()
modelo.fit(horas, calificacion)

#Pendiente y intercepto
print("Pendiente (m): ", modelo.coef_[0])
print("Intercepto (b): ", modelo.intercept_)
print("Ecuacion: y =", modelo.coef_[0], "x +", modelo.intercept_)

#Coeficiente de correlacion
correlacion = grupo1['HorasEstudio'].corr(grupo1['Calificacion'])
print("Correlacion de Pearson: ", correlacion)

#Grafica de dispersion con la linea de regresion
plt.figure(figsize=(6,4))
plt.scatter(horas, calificacion, color='blue', label='Datos')
plt.plot(horas, modelo.predict(horas), color='red', label='Linea de regresion')
plt.title('Horas de Estudio vs Calificacion')
plt.xlabel('Horas de Estudio')
plt.ylabel('Calificacion')
plt.legend()
plt.grid(True)
plt.show()
```
