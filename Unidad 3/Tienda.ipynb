```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.array([0, 1, 2, 3, 4, 5])
p = np.array([0.05, 0.10, 0.25, 0.30, 0.20, 0.10])

# 1. Verificar que las probabilidades sumen 1
suma = p.sum()
print("Suma de probabilidades =", round(suma, 10))
print("¿Suma 1? ->", np.isclose(suma, 1.0))
print("¿Todas entre 0 y 1? ->", bool(np.all((p >= 0) & (p <= 1))))

# 2. Distribución de probabilidad
dist = pd.DataFrame({"Unidades vendidas (X)": x, "P(X = x)": p})
display(dist)

plt.figure(figsize=(7, 4))
plt.bar(x, p, color="steelblue", edgecolor="black")
plt.xlabel("Unidades vendidas (X)")
plt.ylabel("P(X = x)")
plt.title("Distribución de probabilidad de X")
plt.xticks(x)
plt.grid(axis="y", alpha=0.3)
plt.show()

# 3. Valor esperado, varianza y desviación estándar
E_X = np.sum(x * p)
E_X2 = np.sum(x**2 * p)
var = E_X2 - E_X**2
sigma = np.sqrt(var)
cv = sigma / E_X

print(f"E(X)   = {E_X:.4f}")
print(f"E(X^2) = {E_X2:.4f}")
print(f"Var(X) = {var:.4f}")
print(f"DE(X)  = {sigma:.4f}")
print(f"Coeficiente de variación = {cv:.4f} ({cv*100:.1f}%)")

# 4. Función de distribución acumulada F(x)
F = np.cumsum(p)
tabla_F = pd.DataFrame({"x": x, "P(X = x)": p, "F(x) = P(X <= x)": F})
display(tabla_F)

plt.figure(figsize=(8, 4))
xs = np.concatenate(([-1], x, [6]))
Fs = np.concatenate(([0], F))
plt.step(xs, np.concatenate((Fs, [1.0]))[:len(xs)], where="post", color="darkred")
plt.scatter(x, F, color="darkred", zorder=3)
plt.scatter(x, np.concatenate(([0], F[:-1])), facecolors="none", edgecolors="darkred", zorder=3)
plt.xlabel("x")
plt.ylabel("F(x)")
plt.title("Función de distribución acumulada F(x)")
plt.xticks(x)
plt.ylim(-0.05, 1.1)
plt.grid(alpha=0.3)
plt.show()
```
