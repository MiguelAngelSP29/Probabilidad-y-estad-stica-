###**Funcion de distribucion acumulada(FDA)**

```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats
```

```python
# Valores de la variable aleatoria
x = [5,6,7,8,9,10]
p = [0.05,0.1,0.2,0.3,0.025,0.10,]

#Calcular funcion acumulada.
F = np.cumsum(p)

#Mostrar resultados
for xi,fi in zip(x, F):
  print(f"F({xi}) = {fi:.2f}")

#Graficar
plt.step(x, F, where='post')
plt.title("Funcion de distribucion acumulada")
plt.xlabel("Calificacion")
plt.ylabel("Probalidad acumulada")
plt.grid(True)
plt.ylim(0,1.05)
plt.show()
```
