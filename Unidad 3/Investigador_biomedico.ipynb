```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python

x = np.array([0, 1, 2, 3, 4, 5])
p = np.array([0.10, 0.25, 0.30, 0.20, 0.10, 0.05])

#Verificar que las probabilidades sumen 1
print("Suma de probabilidades:", round(p.sum(), 10))
print("¿Suma 1?", np.isclose(p.sum(), 1))

#Distribución de probabilidad
dist = pd.DataFrame({"x": x, "P(X=x)": p})
print("\nDistribución de probabilidad:")
print(dist.to_string(index=False))

#Valor esperado, varianza y desviación estándar
E = np.sum(x * p)
E2 = np.sum(x**2 * p)
var = E2 - E**2
sd = np.sqrt(var)
print(f"\nE(X)   = {E:.4f}")
print(f"E(X²)  = {E2:.4f}")
print(f"Var(X) = {var:.4f}")
print(f"σ      = {sd:.4f}")

#Función de distribución acumulada
F = np.cumsum(p)
acum = pd.DataFrame({"x": x, "F(x)": F})
print("\nFunción acumulada:")
print(acum.to_string(index=False))

#Gráficas
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

ax[0].bar(x, p, color="steelblue")
ax[0].set_title("Distribución de probabilidad de X")
ax[0].set_xlabel("Células anormales (x)")
ax[0].set_ylabel("P(X = x)")

#Escalón de F(x)
xs = np.concatenate(([x[0] - 1], x, [x[-1] + 1]))
Fs = np.concatenate(([0], F, [1]))
ax[1].step(xs, Fs, where="post", color="darkred")
ax[1].scatter(x, F, color="darkred", zorder=3)
ax[1].set_title("Función de distribución acumulada F(x)")
ax[1].set_xlabel("Células anormales (x)")
ax[1].set_ylabel("F(x)")
ax[1].set_ylim(0, 1.05)
ax[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
